```python
import pandas as pd
import numpy as np

# Load dataset
df = pd.read_csv("train.csv")

# 1. Remove duplicate rows
df = df.drop_duplicates()

# 2. Convert Date into datetime
df["Date"] = pd.to_datetime(
    df["Date"],
    format="%d-%m-%Y",
    errors="coerce"
)

# 3. Remove rows where Date is invalid/missing
df = df.dropna(subset=["Date"])

# 4. Remove rows where Weekly_Sales is missing
df = df.dropna(subset=["Weekly_Sales"])

# 5. Remove invalid sales values
df = df[df["Weekly_Sales"] >= 0]

# 6. Aggregate sales across all stores and departments
weekly_sales = (
    df.groupby("Date", as_index=False)["Weekly_Sales"]
      .sum()
      .sort_values("Date")
)

# 7. Check final dataset
print("Rows:", len(weekly_sales))
print("Columns:", weekly_sales.columns.tolist())
print("\nMissing values:")
print(weekly_sales.isnull().sum())

print("\nFirst 5 rows:")
display(weekly_sales.head())

# 8. Save cleaned dataset
weekly_sales.to_csv("walmart_cleaned_timeseries.csv", index=False)
```

```python
import pandas as pd

# Read the cleaned file
clean = pd.read_csv("walmart_cleaned_timeseries.csv")

# Force Weekly_Sales to numeric
clean["Weekly_Sales"] = pd.to_numeric(
    clean["Weekly_Sales"],
    errors="coerce"
)

# Convert Date properly
clean["Date"] = pd.to_datetime(clean["Date"])

# Remove any accidental missing values
clean = clean.dropna(subset=["Date", "Weekly_Sales"])

# Sort chronologically
clean = clean.sort_values("Date")

# Save
clean.to_csv(
    "walmart_final_timeseries.csv",
    index=False
)

print(clean.dtypes)
print(clean.shape)
display(clean.head())
```
